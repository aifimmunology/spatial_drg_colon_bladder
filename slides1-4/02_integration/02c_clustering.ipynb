## Neighbors, UMAP, clustering

```python
sc.pp.neighbors(adata, use_rep = 'X_scVI_ModelA_475', random_state = 0)
sc.pp.neighbors(bdata, use_rep = 'X_scVI_ModelB_480', random_state = 0)
```

```python
sc.tl.umap(adata, random_state = 0)
sc.tl.umap(bdata, random_state = 0)
```

```python
sc.pl.umap(adata)
sc.pl.umap(bdata)
```

```python
sc.tl.leiden(adata, key_added='leiden_scVI_1', resolution=1)
sc.tl.leiden(bdata, key_added='leiden_scVI_1', resolution=1)
```

```python
for data in [adata, bdata]:
    sc.pl.umap(data, color=['leiden_scVI_1'], legend_fontsize=10, frameon=False)
    sc.pl.umap(data, color=['sample_id'],     legend_fontsize=10, frameon=False)
    sc.pl.umap(data, color=['slide_id'],         legend_fontsize=10, frameon=False)
```

### Run DE rankings on clusters

```python
# performance warning here but should be fine

sc.tl.rank_genes_groups(adata, groupby='leiden_scVI_1', method='wilcoxon')
df = sc.get.rank_genes_groups_df(adata, group = None)
df = df.copy()
df = df[(df.pvals_adj < 0.05) & (df.logfoldchanges > .5)].copy()
df.head(5)
```

```python
clusters_path = os.path.join(clusters_dir, 'rank-genes-clusters.csv')

df.to_csv(clusters_path, index=False)
print(clusters_path)
```

```python
# -----------------------------------------------
# 1. Define the signature -- extracted from single cell sequencing dataset subsetted to panel genes
# -----------------------------------------------
signature = [
    "Tafa1",
    "Sema5a",
    "Trappc3",   # corrected from "Trapp3"
    "Bmpr1b",
    "Rasgrf2"    # corrected from "Rasgrl2"
]

# -----------------------------------------------
# 2. Keep only genes present in the Xenium dataset
# -----------------------------------------------
sig_present = [g for g in signature if g in adata.var_names]

print("Genes found in Xenium dataset:")
print(sig_present)

if len(sig_present) == 0:
    raise ValueError("None of the signature genes are present in adata.var_names!")

# -----------------------------------------------
# 3. Score cells using Scanpy
# -----------------------------------------------
sc.tl.score_genes(
    adata,
    gene_list=sig_present,
    score_name="dual_proj_score"
)

print("Added signature score to adata.obs['dual_proj_score'].")

# -----------------------------------------------
# 4. Plot on UMAP
# -----------------------------------------------
sc.pl.umap(
    adata,
    color="dual_proj_score",
    #cmap="viridis",
    frameon=False,
    #color_map="viridis",
    #size=20,
    title="Dual Innervating Signature Projection"
)
```
