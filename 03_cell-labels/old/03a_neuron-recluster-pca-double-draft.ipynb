```python
import os
from pathlib import Path
import scanpy as sc
from scipy.sparse import issparse
import pandas as pd
import numpy as np
import anndata as ad
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
import session_info
import sys
```

```python
base_dir = Path('/home/workspace/projects/drg')

input_dir = base_dir / "data/h5ad/export_03/01_broad-labels"
output_dir = base_dir / "data/h5ad/export_03/02_neuron-labels"

#output_dir.mkdir(parents=True, exist_ok=True)
```

## Read data

```python
adata = sc.read_h5ad(os.path.join(input_dir, 'ModelB_whole_noTG_neurons.h5ad'))
adata
```

```python
adata.var
```

## Subset adata.var

```python
cols = ["zhen_list", "neurons_supervised", "neurons_unsupervised", "neuron_DEGs"]

truthy_values = {"true", "t", "yes", "y", "1"}

for c in cols:
    adata.var[c] = (
        adata.var[c]
        .astype(str)
        .str.strip()
        .str.lower()
        .isin(truthy_values)
    )
```

```python
for c in cols:
    print(f"{c}: {adata.var[c].sum()} True")
```

```python
cols = ["zhen_list", "neurons_supervised", "neurons_unsupervised", "neuron_DEGs"]

mask = adata.var[cols].any(axis=1)

genes_keep = adata.var_names[mask]
print(f"Selected {mask.sum()} genes")

adata_subset = adata[:, mask].copy()
adata_subset
```

```python
adata_subset.var
```

# PCA

```python
adata_subset.X = adata_subset.layers['log1p'].copy()
```

```python
# Step 1: Normalize and scale
sc.pp.scale(adata_subset, max_value=10)

# Step 2: Run PCA
sc.tl.pca(adata_subset, svd_solver='arpack')

# Optional: Visualize explained variance
sc.pl.pca_variance_ratio(adata_subset, log=True)
```

```python
sc.pp.neighbors(adata_subset, n_pcs=20)
sc.tl.umap(adata_subset)
```

```python
sc.tl.leiden(adata_subset, key_added="neuron_recluster", resolution=0.4)
```

```python
adata_subset
```

```python
sc.pl.umap(adata_subset, color = 'neuron_recluster', frameon = False, legend_loc = 'on data', title = 'Neuron clusters')
```

```python
# List of barcoded neurons
dual_barcodes = [
    'bpoikicj-1-1',
    'akcoadof-1-2',
    'bgmdeeeb-1-2',
    'hjjgalia-1-2',
    'bmjgagok-1-3',
    'ijnggkoc-1-3'
]

# Make sure these exist in adata_subset
dual_mask = adata_subset.obs_names.isin(dual_barcodes)

print("Found in adata_subset:", adata_subset.obs_names[dual_mask].tolist())

# Convert barcodes to integer indices
dual_indices = [adata_subset.obs_names.get_loc(cell) for cell in adata_subset.obs_names[dual_mask]]

# UMAP coordinates
umap = adata_subset.obsm["X_umap"]

# Visualization
plt.figure(figsize=(5,5), dpi=140)

# Background cells
plt.scatter(
    umap[:, 0],
    umap[:, 1],
    c="lightgray",
    s=5, alpha=0.25,
    label="All cells"
)

# Highlight dual-barcoded neurons
plt.scatter(
    umap[dual_indices, 0],
    umap[dual_indices, 1],
    s=60,
    facecolors=(1, 0, 0, 0.45),  # translucent red fill
    edgecolors="black",
    linewidths=1.2,
    marker="o",
    label="Dual-barcoded neurons"
)

plt.title("Dual-barcoded neurons in adata_subset UMAP")
plt.xticks([]); plt.yticks([])
plt.legend(frameon=False)
sns.despine(left=True, bottom=True)
plt.tight_layout()
plt.show()
```

```python
sc.tl.rank_genes_groups(adata_subset, groupby="neuron_recluster", method="wilcoxon", use_raw = False)
sc.pl.rank_genes_groups(adata_subset, groupby="neuron_recluster", n_genes=30, size=8, use_raw = False)
```

```python
plt.figure(figsize=(7,5), dpi=140)

sns.histplot(
    data=adata_subset.obs,
    x="total_counts",
    bins=50,
    kde=True,
    color="steelblue",
    alpha=0.7
)

plt.xlabel("Total counts per cell")
plt.ylabel("Number of cells")
plt.title("Distribution of Total Transcript Counts")
sns.despine()
plt.tight_layout()
plt.show()
```

```python
sc.pl.umap(adata_subset, color = ['total_counts', 'Htr3a', 'Mrgprd', 'Sst', 'Tac1', 'Trpv1', 'Il31ra', 'Pvalb', 'Scn10a', 'Slc17a6', 'Slc17a7'])
```

```python
genes = ['Nefl', 'Scn10a', 'Cysltr2', 'Il31ra', 'Sst', 'Pvalb', 'Trpv1', 'Htr3a', 'Tac1', 'Ache', 'Thy1', 'Bmpr1b', 'Piezo2', 'Prkcb']
sc.tl.dendrogram(adata_subset, groupby = 'neuron_recluster')
sc.pl.matrixplot(adata_subset, var_names = genes, groupby = 'neuron_recluster', standard_scale = 'var', dendrogram = True)
```

```python
cluster_key = "neuron_recluster"
cluster_ids = sorted(adata_subset.obs[cluster_key].unique())

plt.figure(figsize=(9, 6))

# KDE per cluster
for clust in cluster_ids:
    sns.kdeplot(
        adata_subset.obs.loc[adata_subset.obs[cluster_key] == clust, 'cell_area'],
        fill=False, alpha=0.6, linewidth=1.3,
        label=f"Cluster {clust}"
    )

# Dual-innervated cell areas in adata_subset
dual_mask_subset = adata_subset.obs_names.isin(dual_barcodes)
dual_areas = adata_subset.obs.loc[dual_mask_subset, 'cell_area']

# Mark dual-innervated cells with vertical dotted lines
for x in dual_areas:
    plt.axvline(x, color="red", linestyle="--", linewidth=1.2, alpha=0.85)

plt.title("Cell Area Distribution by Cluster\n(dual-innervated neurons in red)")
plt.xlabel("Cell area")
plt.ylabel("Density")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', frameon=False)
plt.tight_layout()
plt.show()
```

```python
cluster_key = "neuron_recluster"

# Extract relevant data
plot_data = adata_subset.obs[[cluster_key, "cell_area"]].copy()
plot_data[cluster_key] = plot_data[cluster_key].astype(str)

# Dual mask for subset
dual_mask_subset = adata_subset.obs_names.isin(dual_barcodes)

plt.figure(figsize=(11, 6), dpi=150)

# Violin plot (half split, no inner points to reduce clutter)
sns.violinplot(
    data=plot_data,
    x=cluster_key,
    y="cell_area",
    inner=None,
    cut=0,
    linewidth=1.0,
    palette="pastel",
)

# Overlay dual-innervated cells
sns.stripplot(
    data=adata_subset.obs.loc[dual_mask_subset],
    x=cluster_key,
    y="cell_area",
    color="red",
    size=10,
    edgecolor="black",
    linewidth=1.1,
    jitter=True,
    marker="o",
    label="Dual-innervated neurons"
)

plt.title("Cell Area by Cluster\n(red = dual-innervated neurons)", fontsize=14, weight='bold')
plt.xlabel("Cluster", fontsize=12)
plt.ylabel("Cell area", fontsize=12)

sns.despine()
plt.legend(frameon=False, loc="upper right")
plt.tight_layout()
plt.show()
```
