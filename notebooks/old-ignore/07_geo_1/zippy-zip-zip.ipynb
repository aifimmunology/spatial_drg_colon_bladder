# Zip files for upload to project store

```python
import shutil
from pathlib import Path
```

```python
# Define base paths
dest_dir = Path("/home/workspace/private/geo_organized")
zip_base = dest_dir.parent / "drg_geo_organized"
```

```python
print(f"Zipping directory: {dest_dir}")

archive_path = shutil.make_archive(
    base_name=str(zip_base),  
    format="zip",
    root_dir=str(dest_dir.parent),
    base_dir=dest_dir.name
)

print(f"\nArchive created: {archive_path}")
```
